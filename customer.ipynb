```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x:x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df.columes=df.columns.str.lower()
df.columes=df.columns.str.replace(' ','_')
df=df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
print(df.columns)
```

```python
#create a colume age_group
labels = ['Young Adult', 'Adult', 'Middle-aged', 'Senior']
df['Age_group'] = pd.qcut(df['Age'], q=4, labels = labels)
```

```python
df[['Age','Age_group']].head(10)
```

```python
print(df.columns)
```

```python
# create column purches _frequency_days

frequency_mapping = {
    'Fortnightly':14,
    'Weekly':7,
    'Monthly':30,
    'Quaryerly':90,
    'Bi-Weekly':14,
    'Annually':365,
    'Every 3 Months' :90
}
df['Purchase_Frequency_days']=df['Frequency of Purchases'].map(frequency_mapping)
```

```python
df[['Purchase_Frequency_days' , 'Frequency of Purchases']].head(10)
```
